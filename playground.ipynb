```python
import pandas as pd
import numpy as np
```

```python
fibonacci_start = [0, 1]

def get_fibonacci(n):
    if n < len(fibonacci_start):
        return fibonacci_start[n]
    else:
        return get_fibonacci(n-1) + get_fibonacci(n-2)

```

```python
get_fibonacci(4)
```

```python
i = 0
res = 0
while res < 11112025:
    res = get_fibonacci(i)
    i += 1

res, i
```
